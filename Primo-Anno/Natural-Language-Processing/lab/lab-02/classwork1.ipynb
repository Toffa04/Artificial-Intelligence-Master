## Gutenberg Classworks

- We are going to use **Jane Austen's Emma**

```python
import nltk

nltk.download("gutenberg")
nltk.download("punkt_tab")
nltk.download("stopwords")
```

Let's load the **corpus** dedicated

```python
from nltk.corpus import gutenberg 

print(gutenberg.fileids())
```

Now load the text of the **corpus** 

```python
raw_text = gutenberg.raw("austen-emma.txt")

print(len(raw_text))
print(raw_text[:300])
```

Tokeninzation

```python
tokens = nltk.word_tokenize(raw_text)

print(len(tokens))
print(tokens[:40])
```

Now we are going to tokenize the document and convert alphabetic tokens to **lowercase**

```python
words = []

for token in tokens:
    if token.isalpha():
        words.append(token.lower())
       

print(len(words))
print(words[:30])
```

Constructing the **Bag-of-Words representation**

```python
from collections import Counter 

bow = Counter(words)

print("Total number of tokens:", len(words))
print("Vocabulary size:", len(bow))
#print("Frequencies of the 20 most frequent word types:", bow.most_common(20))

for word, count in bow.most_common(20):
    print(word, count)
```

Now remove the English stopwords and construct a **second BoW representation**

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

print(len(stop_words))
print(list(stop_words)[:10])
```

```python
filtered_words = []

for word in words:
    if word not in stop_words:
        filtered_words.append(word)

print(len(words), len(filtered_words))
print(filtered_words[:30])
```

```python
bow_filtered = Counter(filtered_words)

print("Total number of token:", len(filtered_words))
print("Vocabulary size:", len(bow_filtered))

for word, count in bow_filtered.most_common(20):
    print(word, count)
```

## Bag-of-Words on Jane Austen's Emma: effect of preprocessing

**Objective**

Build a Bag-of-Words (BoW) representation of *Emma* NLTK Gutenberg corpus, (`austen-emma.txt`) twice, once before and once after English stopword removal, and compare the two to see how much preprocessing decisions shape the information a BoW conveys.

## Method
 
1. Load the raw text with `gutenberg.raw("austen-emma.txt")` (887,071 characters).
2. Tokenize with `nltk.word_tokenize`.
3. Keep only alphabetic tokens (`isalpha()`) and convert them to lowercase.
4. Build BoW 1 with `collections.Counter` and report total tokens, vocabulary size and the 20 most frequent word types.
5. Remove English stopwords (`nltk.corpus.stopwords`, stored as a `set`) and build BoW 2 with the same statistics.
6. Compare the two representations.
## Results
 
| | BoW 1 (no stopword removal) | BoW 2 (stopwords removed) | Change |
|---|---|---|---|
| Total tokens | 157,114 | 69,695 | -87,419 (about -56%) |
| Vocabulary size | 6,933 | 6,809 | -124 (about -1.8%) |
 
### Top 20 words
 
**Before stopword removal:** the (5201), to (5181), and (4877), of (4284), i (3177), a (3124), it (2503), her (2448), was (2396), she (2336), not (2281), in (2173), be (1970), you (1967), he (1806), that (1805), had (1623), but (1441), as (1436), for (1346).
 
**After stopword removal:** emma (860), could (836), would (818), miss (599), must (566), harriet (500), much (484), said (483), one (447), weston (437), every (435), thing (394), think (383), elton (383), knightley (379), well (378), little (359), never (358), know (335), might (325).
 
## Discussion
 
**Before preprocessing**, the 20 most frequent words are all function words. This list describes the English language rather than the novel: almost the same list would appear for any other English text. Note also that `i` mixes the pronoun with the Roman numerals in the headings `VOLUME I` and `CHAPTER I`.
 
**After stopword removal**, the picture changes. None of the 20 words from the first list survives. The most characteristic terms are now the main characters (`emma`, `harriet`, `weston`, `elton`, `knightley`) and the title `miss`, which point directly to the story and its social setting.
 
**Tokens vs. vocabulary.** Stopwords are very few as word types (only 124 of them appear in this text, under 2% of the vocabulary) but very frequent as occurrences: removing them eliminated more than half of all tokens. A small group of words dominates the counts.
 
**Remaining noise.** The standard NLTK stopword list does not remove modal verbs (`could`, `would`, `must`, `might`) or generic words (`said`, `one`, `much`, `every`, `thing`), which remain among the most frequent words but say little about the novel. A stopword list is therefore a choice, not a neutral filter.
 
## Answer to the main question
 
**How strongly does preprocessing affect the information conveyed by a Bag-of-Words?**
 
Very strongly. The same document produced two BoWs whose top-20 lists have no word in common, and the information they carry is of a different kind: language structure in the first case, characters and content in the second. The decisions made before counting also change the result in other ways:
 
- the tokenizer determines what counts as a token;
- keeping only alphabetic tokens discards words containing punctuation or hyphens (for example `twenty-one`) as well as numbers;
- lowercasing merges `The` and `the` into one type;
- the stopword list decides which frequent words are considered uninformative;
- headings and Roman numerals stay in the text unless removed explicitly.
A BoW is therefore not simply determined by the source text: it also depends on the choices made before the counts are built.
